```python
import os
import numpy as np
import pandas as pd
import nibabel as nib
# import scipy.io 
# import h5py
```

```python
def export_cifti(outpath, data, headers_list=None, verbose=True):
    if "/Users/snaranjo" in os.getcwd():
        local_pc_flag = "/Users/snaranjo/Desktop/neurotranslate/mount_point"
    else:
        local_pc_flag=""
    if verbose:
        print("exporting to CIFTI2 format...")
    ext = os.path.splitext(os.path.basename(outpath))[1]
    outpath = outpath.replace(ext, '.dscalar.nii')
    if not headers_list:
        #eaxmple header for use
        # melodic_dscalar_example=local_pc_flag+'/ceph/chpc/shared/janine_bijsterbosch_group/naranjorincon_scratch/NeuroTranslate/generate_ICA/ABCD_ICA/ICAd15/melodic_IC.nii'
        melodic_dscalar_example=local_pc_flag+'/ceph/chpc/shared/janine_bijsterbosch_group/WAPIAW_2026/ICA/ABCD_ICAd14_4mm/melodic_IC.nii'
        cifti_data = nib.load(melodic_dscalar_example)
        headers_list = [cifti_data.header, cifti_data.nifti_header]
    
    header = headers_list[0]
    nifti_header = headers_list[1]
    
    # unpack CIFTI2 axes, assuming len(data.shape)==2
    _, brain_axis = [header.get_axis(i) for i in range(len(data.shape))]

    # update cifti header
    scalar_axis=nib.cifti2.ScalarAxis(['']*20) #20 hard coded here cause i know infomap is 20 nets
    new_header = nib.Cifti2Header.from_axes([scalar_axis, brain_axis])
    # create cifti template data
    new_cdata = nib.Cifti2Image(data, header=new_header, nifti_header=nifti_header)
    print("cifti data shape: " + str(new_cdata.shape))
    # export data to cifti format
    new_cdata.to_filename(outpath)
    if verbose:
        print("CIFTI2 data saved to " + outpath)

def _zero_pad(data, n_rows=91282):
    zero_stack = np.zeros(data.shape[:-1] + (n_rows-data.shape[-1],))
    padded_data = np.append(data, zero_stack, axis=-1)
    return padded_data
```

```python
if "/Users/snaranjo" in os.getcwd():
    local_pc_flag = "/Users/snaranjo/Desktop/neurotranslate/mount_point"
else:
    local_pc_flag=""
    
# get priors
shared_path=local_pc_flag+"/ceph/chpc/shared/janine_bijsterbosch_group/naranjorincon_scratch"
lynch_path_priors_dir=f"{shared_path}/NeuroTranslate/surf2netmat/infomap_20net_prior"
get_spatial_priors=nib.load(f"{lynch_path_priors_dir}/infomap_20_spatial_prior.dtseries.nii").get_fdata()
output_path=f"{lynch_path_priors_dir}/infomap_spatial_priors.nii" #will add the dscalar
export_cifti(output_path, get_spatial_priors)

get_func_priors=nib.load(f"{lynch_path_priors_dir}/infomap_20_fc_prior.dtseries.nii").get_fdata()
output_path=f"{lynch_path_priors_dir}/infomap_FC_priors.nii"
export_cifti(output_path, get_func_priors)
```
